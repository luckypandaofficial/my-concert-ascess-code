### Checking Three Conditions Together
We can check multiple criteria using nested conditions or multiple logical operators (`and`, `or`, `not`). Here, we check age, ticket ownership, and VIP status.

```python
age = int(input("Enter your age: "))
has_ticket = True
is_vip = True

if age >= 18 and has_ticket and is_vip:
    print("Welcome! You can enter the concert and access the VIP lounge.")
elif age >= 18 and has_ticket:
    print("You can enter the concert, but VIP access is denied.")
else:
    print("Access denied.")
```

### Validating with Real Ticket Codes

```python
# A list of valid ticket codes in our database
valid_tickets = ["TKT-1012", "TKT-3045", "VIP-9912", "VIP-7788"]

# Take input from the user
user_ticket = input("Please enter your ticket code (e.g., TKT-1012): ").strip().upper()

# Verify if the ticket is in our list of valid tickets
if user_ticket in valid_tickets:
    print(f"Ticket '{user_ticket}' verified! Access granted.")
else:
    print(f"Ticket '{user_ticket}' is invalid or has already been used. Access denied.")
```

```python
# 1. Define valid ticket database
valid_tickets = ["TKT-1012", "TKT-3045", "VIP-9912", "VIP-7788"]

# 2. Collect user inputs
age = int(input("Enter your age: "))
user_ticket = input("Please enter your ticket code (e.g., TKT-1012): ").strip().upper()

# 3. Perform Ticket Validation
has_ticket = user_ticket in valid_tickets
is_vip = user_ticket.startswith("VIP-") and has_ticket

# 4. Determine Entry Eligibility
if not has_ticket:
    print(f"Ticket '{user_ticket}' is invalid or has already been used. Access denied.")
elif age >= 18 and is_vip:
    print(f"Ticket '{user_ticket}' verified! Welcome! You can enter the concert and access the VIP lounge.")
elif age >= 18:
    print(f"Ticket '{user_ticket}' verified! You can enter the concert, but VIP access is denied.")
else:
    print("Access denied. You must be at least 18 years old to enter.")
```

### Explicit Verification: Normal vs. VIP Entry

```python
# 1. Our active ticket system databases
normal_tickets = ["TKT-1012", "TKT-3045"]
vip_tickets = ["VIP-9912", "VIP-7788"]

# 2. Get user age and ticket code
age = int(input("Enter your age: "))
user_ticket = input("Enter your ticket code (e.g. VIP-9912 or TKT-1012): ").strip().upper()

# 3. Process Ticket Type
is_valid_vip = user_ticket in vip_tickets
is_valid_normal = user_ticket in normal_tickets

# 4. Check Access Level
if age < 18:
    print("Access denied. You must be at least 18 years old to enter.")
elif is_valid_vip:
    print(f"Ticket '{user_ticket}' Verified! You have VIP ENTRY. Access granted to both the concert and the VIP lounge.")
elif is_valid_normal:
    print(f"Ticket '{user_ticket}' Verified! You have NORMAL ENTRY. Access granted to the concert (VIP lounge restricted).")
else:
    print(f"Ticket '{user_ticket}' is invalid. Access denied.")
```
